## Paso 4 · Errores de interpretación que hay que evitar

El documento advierte: *"Un cálculo correcto con una interpretación incorrecta no obtiene la totalidad de los puntos."* Estas son las frases incorrectas más comunes y cómo corregirlas:

| ❌ No decir | ✅ Decir |
|---|---|
| "El abandono es 0.206." | "La proporción de abandono **en la muestra** fue 0.206; es una estimación del abandono de los 10,000 clientes." |
| "Hay 95 % de probabilidad de que el abandono esté entre 0.182 y 0.230." | "Con 95 % de confianza, el abandono está entre 0.182 y 0.230. El 95 % es la tasa de acierto del método." |
| "La inactividad causa el abandono." | "Los inactivos abandonan más; el estudio es observacional y no prueba la causa." |
| "Como no se rechaza H0, la muestra es igual a la población." | "No hay evidencia de que la muestra sea distinta de la población." |
| "El intervalo de P4 falló, así que el método está mal." | "Con 95 % de confianza se espera que cerca de 1 de cada 20 intervalos falle; la conclusión de P4 se mantiene." |
| "Los resultados aplican a los bancos europeos." | "Los resultados aplican a esta cartera de 10,000 clientes." |
| "Usamos 95 % porque así se hace." | "Usamos 95 % porque equilibra seguridad y precisión, y los errores de esta decisión son económicos y corregibles." |
